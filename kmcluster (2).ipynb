## K means cluster

##### identifies k cluster(grouping) within the dataset

##### Steps :

> 1:Choose 3(based on k value)  random points as  Centroids(mean)

> 2:Calculate distance between points and centroids and assigne points to closest centroid

> 3:compute new centroids ..

NOTE :  we need to repeat step 2 and step 3 until we reach stability (centroids value gets repeated)

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
iris.data
```

```python
columns = iris.feature_names
```

```python
columns 
```

```python
df = pd.DataFrame(iris.data, columns=columns)
```

```python
df
```

```python
### Splitting as training and testing data
from sklearn.model_selection import train_test_split

train ,test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.shape
```

```python
test.shape
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3, random_state=42)
```

```python
model.fit(train)
```

```python
predict = model.predict(test)
```

```python
pd.Series(predict).unique()
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
sns.scatterplot(x=train['petal length (cm)'], y=train['petal width (cm)'], hue=model.labels_, palette='Set1')
```

```python
sns.scatterplot(x=test['petal length (cm)'], y=test['petal width (cm)'], hue=predict, palette='Set1')
```
